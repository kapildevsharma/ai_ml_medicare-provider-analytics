This notebook runs the **complete EDA code directly in notebook cells**. It does not create a separate Python script or use `subprocess.run()`. The workflow is EDA only; it does not train or validate predictive models.

```python
# ================================================================
# QM640 DATA ANALYTICS CAPSTONE
# Medicare Provider Clustering and Cluster-Wise Hypothesis Testing
#
# Project:
# Predicting Medicare Payment Costs and Identifying High-Cost
# Provider Patterns Using Machine Learning
#
# Dataset: CMS Medicare Physician & Other Practitioners - 2024
# Target: Tot_Mdcr_Pymt_Amt
#
# IMPORTANT:
# Medicare payment is excluded from clustering features.
# It is used only after clustering for profiling and hypothesis tests.
# ================================================================


# ================================================================
# STEP 1: INSTALL AND IMPORT LIBRARIES
# ================================================================

!pip -q install scikit-learn scipy pandas numpy matplotlib


from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import spearmanr, kruskal

from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")


# ================================================================
# STEP 2: MOUNT GOOGLE DRIVE
# ================================================================

from google.colab import drive

drive.mount("/content/drive")


# ================================================================
# STEP 3: DEFINE PROJECT PATHS AND SETTINGS
# ================================================================

PROJECT_DIR = Path(
    "/content/drive/MyDrive/AI_ML-MS/Capstone Project/qm640_project"
)

INPUT_FILE = PROJECT_DIR / "cms_provider_2024.csv"

OUTPUT_DIR = PROJECT_DIR / "outputs" / "clustering"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

TARGET = "Tot_Mdcr_Pymt_Amt"

# Candidate features for clustering.
# These must exist in the actual CMS dataset.
CANDIDATE_FEATURES = [
    "Tot_Benes",
    "Tot_Srvcs",
    "Tot_Bene_Day_Srvcs",
]

# Categorical variables used to describe clusters.
PROFILE_CATEGORICAL_COLUMNS = [
    "Rndrng_Prvdr_Type",
    "Rndrng_Prvdr_State_Abrvtn",
]

# Candidate number of clusters.
K_VALUES = range(2, 7)

# Reproducibility.
RANDOM_STATE = 42

# Computational sample settings.
SAMPLE_FRAC = 0.10
MAX_SAMPLE_ROWS = 150_000

# Hypothesis-testing settings.
ALPHA = 0.05

print("Project directory:", PROJECT_DIR)
print("Input file:", INPUT_FILE)
print("Output directory:", OUTPUT_DIR)

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Dataset not found: {INPUT_FILE}\n"
        "Check the Google Drive path and CSV filename."
    )


# ================================================================
# STEP 4: INSPECT DATASET SCHEMA
# ================================================================

header = pd.read_csv(INPUT_FILE, nrows=0).columns.tolist()

print("\nNumber of columns:", len(header))
print("\nAvailable columns:")
print(header)

if TARGET not in header:
    raise ValueError(
        f"Target column '{TARGET}' was not found. "
        "Verify the column name in the CMS CSV."
    )

CLUSTER_FEATURES = [
    col for col in CANDIDATE_FEATURES
    if col in header
]

PROFILE_CATEGORICAL_COLUMNS = [
    col for col in PROFILE_CATEGORICAL_COLUMNS
    if col in header
]

IDENTIFIER_COLUMNS = [
    col for col in ["Rndrng_NPI", "NPI"]
    if col in header
]

if len(CLUSTER_FEATURES) < 2:
    raise ValueError(
        "At least two candidate clustering features must be present. "
        f"Available features: {CLUSTER_FEATURES}"
    )

print("\nClustering features:", CLUSTER_FEATURES)
print("Profiling categories:", PROFILE_CATEGORICAL_COLUMNS)
print("Identifier columns:", IDENTIFIER_COLUMNS)


# ================================================================
# STEP 5: LOAD REQUIRED COLUMNS
# ================================================================

USE_COLUMNS = list(
    dict.fromkeys(
        IDENTIFIER_COLUMNS
        + CLUSTER_FEATURES
        + PROFILE_CATEGORICAL_COLUMNS
        + [TARGET]
    )
)

df = pd.read_csv(
    INPUT_FILE,
    usecols=USE_COLUMNS,
    low_memory=False
)

original_row_count = len(df)

print("\nOriginal dataset rows:", original_row_count)
print("Loaded columns:", df.columns.tolist())

print("\nDataset sample:")
display(df.head())

print("\nData types:")
display(df.dtypes.astype(str).rename("Data Type").to_frame())


# ================================================================
# STEP 6: DATA QUALITY CHECKS
# ================================================================

quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": [str(df[c].dtype) for c in df.columns],
    "Missing_Count": [int(df[c].isna().sum()) for c in df.columns],
    "Missing_Percent": [
        round(df[c].isna().mean() * 100, 3)
        for c in df.columns
    ],
    "Unique_Values": [int(df[c].nunique(dropna=True)) for c in df.columns]
})

quality_report.to_csv(
    OUTPUT_DIR / "clustering_data_quality.csv",
    index=False
)

print("\nData quality report:")
display(quality_report)

print("\nExact duplicate rows:", int(df.duplicated().sum()))

if "Rndrng_NPI" in df.columns:
    print(
        "Duplicate non-null rendering provider NPIs:",
        int(
            df.loc[
                df["Rndrng_NPI"].notna(),
                "Rndrng_NPI"
            ].duplicated().sum()
        )
    )

# Convert numeric fields safely.
for col in CLUSTER_FEATURES + [TARGET]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# The target is needed for the later hypothesis tests.
# Keep only rows with a valid target.
df = df.dropna(subset=[TARGET]).copy()

# Negative service/beneficiary counts are treated as invalid/missing.
# Do not automatically remove extreme but valid values.
for col in CLUSTER_FEATURES:
    df.loc[df[col] < 0, col] = np.nan


# ================================================================
# STEP 7: CREATE A REPRODUCIBLE ANALYSIS SAMPLE
# ================================================================

rows_before_sampling = len(df)

if SAMPLE_FRAC < 1.0:
    df = df.sample(
        frac=SAMPLE_FRAC,
        random_state=RANDOM_STATE
    )

if len(df) > MAX_SAMPLE_ROWS:
    df = df.sample(
        n=MAX_SAMPLE_ROWS,
        random_state=RANDOM_STATE
    )

df = df.reset_index(drop=True)

print("\nRows with valid target before sampling:", rows_before_sampling)
print("Rows in clustering analysis:", len(df))
print("Requested sampling fraction:", SAMPLE_FRAC)
print("Maximum sample rows:", MAX_SAMPLE_ROWS)

print(
    "\nNOTE: This sample is selected for computational feasibility. "
    "It is not a power-based sample-size calculation."
)


# ================================================================
# STEP 8: PREPARE CLUSTERING FEATURES
# ================================================================

# Remove features that are completely missing or constant.
CLUSTER_FEATURES = [
    col for col in CLUSTER_FEATURES
    if df[col].notna().any()
    and df[col].nunique(dropna=True) > 1
]

if len(CLUSTER_FEATURES) < 2:
    raise ValueError(
        "Fewer than two usable clustering features remain."
    )

X_raw = df[CLUSTER_FEATURES].copy()

# Counts are usually right-skewed.
# log1p reduces skew while preserving zero values.
X_log = np.log1p(X_raw.clip(lower=0))

# Median imputation.
imputer = SimpleImputer(strategy="median")

X_imputed = imputer.fit_transform(X_log)

# Standardization ensures features with large numerical scales
# do not dominate Euclidean-distance-based K-Means.
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_imputed)

print("\nFeatures used for clustering:")
print(CLUSTER_FEATURES)

print("\nScaled feature matrix shape:", X_scaled.shape)


# ================================================================
# STEP 9: SELECT NUMBER OF CLUSTERS USING SILHOUETTE SCORE
# ================================================================

# Silhouette calculations can be expensive on large datasets.
# Use a reproducible subset for calculating silhouette scores.
silhouette_sample_size = min(10_000, len(df))

rng = np.random.RandomState(RANDOM_STATE)

silhouette_indices = rng.choice(
    len(df),
    size=silhouette_sample_size,
    replace=False
)

X_silhouette = X_scaled[silhouette_indices]

silhouette_results = []

for k in K_VALUES:

    if len(df) <= k:
        continue

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    all_labels = model.fit_predict(X_scaled)

    sampled_labels = all_labels[silhouette_indices]

    if len(np.unique(sampled_labels)) < 2:
        score = np.nan
    else:
        score = silhouette_score(
            X_silhouette,
            sampled_labels
        )

    silhouette_results.append({
        "Number_of_Clusters": k,
        "Silhouette_Score": score
    })

silhouette_df = pd.DataFrame(silhouette_results)

silhouette_df.to_csv(
    OUTPUT_DIR / "silhouette_scores.csv",
    index=False
)

print("\nSilhouette results:")
display(silhouette_df)

valid_scores = silhouette_df.dropna()

if valid_scores.empty:
    raise RuntimeError(
        "No valid silhouette scores could be calculated."
    )

best_row = valid_scores.loc[
    valid_scores["Silhouette_Score"].idxmax()
]

BEST_K = int(best_row["Number_of_Clusters"])

print("\nSelected number of clusters:", BEST_K)
print("Selected silhouette score:", best_row["Silhouette_Score"])


# ================================================================
# STEP 10: FIT FINAL K-MEANS MODEL
# ================================================================

final_model = KMeans(
    n_clusters=BEST_K,
    random_state=RANDOM_STATE,
    n_init=10
)

df["Cluster"] = final_model.fit_predict(X_scaled)

print("\nCluster assignment completed.")

print("\nCluster counts:")
display(
    df["Cluster"]
    .value_counts()
    .sort_index()
    .rename_axis("Cluster")
    .reset_index(name="Provider_Count")
)


# ================================================================
# STEP 11: PROFILE EACH CLUSTER
# ================================================================

profile_rows = []

PROFILE_NUMERIC_COLUMNS = list(
    dict.fromkeys(CLUSTER_FEATURES + [TARGET])
)

for cluster_id, group in df.groupby("Cluster"):

    row = {
        "Cluster": int(cluster_id),
        "Provider_Count": int(len(group)),
        "Provider_Percent": round(
            100 * len(group) / len(df),
            3
        )
    }

    # Numerical summaries.
    for col in PROFILE_NUMERIC_COLUMNS:

        row[f"{col}_mean"] = group[col].mean()
        row[f"{col}_median"] = group[col].median()
        row[f"{col}_p90"] = group[col].quantile(0.90)

    # Categorical summaries.
    for col in PROFILE_CATEGORICAL_COLUMNS:

        nonmissing = group[col].dropna()
        mode = nonmissing.mode()

        if len(mode) > 0:
            most_common = mode.iloc[0]

            row[f"{col}_most_common"] = str(most_common)

            row[f"{col}_most_common_percent"] = (
                100 * (nonmissing == most_common).mean()
            )
        else:
            row[f"{col}_most_common"] = "Unknown"
            row[f"{col}_most_common_percent"] = np.nan

    profile_rows.append(row)

cluster_profiles = pd.DataFrame(profile_rows)

cluster_profiles = cluster_profiles.sort_values("Cluster")

cluster_profiles.to_csv(
    OUTPUT_DIR / "cluster_profiles.csv",
    index=False
)

print("\nCLUSTER PROFILES")
display(cluster_profiles)


# ================================================================
# STEP 12: SAVE PROVIDER-TO-CLUSTER ASSIGNMENTS
# ================================================================

assignment_columns = (
    IDENTIFIER_COLUMNS
    + CLUSTER_FEATURES
    + [TARGET, "Cluster"]
)

assignment_columns = [
    col for col in assignment_columns
    if col in df.columns
]

assignments = df[assignment_columns].copy()

assignments.to_csv(
    OUTPUT_DIR / "provider_cluster_assignments.csv",
    index=False
)

print(
    "\nSaved provider assignments:",
    OUTPUT_DIR / "provider_cluster_assignments.csv"
)


# ================================================================
# STEP 13: SAVE CLUSTER SIZE SUMMARY
# ================================================================

cluster_sizes = (
    df["Cluster"]
    .value_counts()
    .sort_index()
    .rename_axis("Cluster")
    .reset_index(name="Provider_Count")
)

cluster_sizes["Provider_Percent"] = (
    100 * cluster_sizes["Provider_Count"] / len(df)
)

cluster_sizes.to_csv(
    OUTPUT_DIR / "cluster_size_summary.csv",
    index=False
)

print("\nCLUSTER SIZE SUMMARY")
display(cluster_sizes)


# ================================================================
# STEP 14: WITHIN-CLUSTER HYPOTHESIS TESTING
# ================================================================

# For every cluster and feature:
#
# H0: There is no monotonic association between the feature
#     and Medicare payments within this cluster.
#
# H1: There is a monotonic association between the feature
#     and Medicare payments within this cluster.
#
# Test: Spearman rank correlation.
#
# H0: rho = 0
# H1: rho != 0
#
# Holm correction is applied across all valid planned tests.

def holm_adjustment(p_values):

    p_values = np.asarray(p_values, dtype=float)

    number_of_tests = len(p_values)

    order = np.argsort(p_values)

    sorted_p = p_values[order]

    adjusted_sorted = np.maximum.accumulate(
        (number_of_tests - np.arange(number_of_tests))
        * sorted_p
    )

    adjusted_sorted = np.minimum(
        adjusted_sorted,
        1.0
    )

    adjusted = np.empty(number_of_tests)

    adjusted[order] = adjusted_sorted

    return adjusted


hypothesis_features = [
    col for col in [
        "Tot_Srvcs",
        "Tot_Benes",
        "Tot_Bene_Day_Srvcs"
    ]
    if col in df.columns
]

hypothesis_results = []

for cluster_id, group in df.groupby("Cluster"):

    print(f"\nTesting Cluster {cluster_id}")

    for feature in hypothesis_features:

        pair = group[
            [feature, TARGET]
        ].replace(
            [np.inf, -np.inf],
            np.nan
        ).dropna()

        n = len(pair)

        if (
            n < 3
            or pair[feature].nunique() < 2
            or pair[TARGET].nunique() < 2
        ):

            hypothesis_results.append({
                "Cluster": int(cluster_id),
                "Feature": feature,
                "N": n,
                "Spearman_Rho": np.nan,
                "P_Value": np.nan,
                "Status": "Not tested: insufficient data or constant variable"
            })

            continue

        rho, p_value = spearmanr(
            pair[feature],
            pair[TARGET]
        )

        hypothesis_results.append({
            "Cluster": int(cluster_id),
            "Feature": feature,
            "N": n,
            "Spearman_Rho": rho,
            "P_Value": p_value,
            "Status": "Tested"
        })


hypothesis_df = pd.DataFrame(hypothesis_results)

# Adjust p-values across all valid tests.
valid_tests = hypothesis_df["P_Value"].notna()

hypothesis_df["Holm_Adjusted_P_Value"] = np.nan

if valid_tests.any():

    hypothesis_df.loc[
        valid_tests,
        "Holm_Adjusted_P_Value"
    ] = holm_adjustment(
        hypothesis_df.loc[
            valid_tests,
            "P_Value"
        ].values
    )

hypothesis_df["Alpha"] = ALPHA

hypothesis_df["Reject_H0"] = pd.NA

hypothesis_df.loc[
    valid_tests,
    "Reject_H0"
] = (
    hypothesis_df.loc[
        valid_tests,
        "Holm_Adjusted_P_Value"
    ] < ALPHA
)


def interpret_test(row):

    if pd.isna(row["P_Value"]):
        return (
            "Not tested because of insufficient data "
            "or a constant variable."
        )

    if not bool(row["Reject_H0"]):
        return (
            "No statistically significant association "
            "after Holm correction."
        )

    if row["Spearman_Rho"] > 0:
        return (
            "Positive monotonic association; statistically "
            "significant after Holm correction."
        )

    if row["Spearman_Rho"] < 0:
        return (
            "Negative monotonic association; statistically "
            "significant after Holm correction."
        )

    return "No observed monotonic association."


hypothesis_df["Interpretation"] = hypothesis_df.apply(
    interpret_test,
    axis=1
)

hypothesis_df.to_csv(
    OUTPUT_DIR / "cluster_hypothesis_tests.csv",
    index=False
)

print("\nCLUSTER-WISE HYPOTHESIS TEST RESULTS")
display(hypothesis_df)


# ================================================================
# STEP 15: OPTIONAL GLOBAL TEST OF PAYMENT DIFFERENCES
# ACROSS CLUSTERS
# ================================================================

# This is different from the within-cluster correlation tests.
#
# H0: Payment distributions are the same across clusters.
# H1: At least one cluster's payment distribution differs.
#
# Kruskal-Wallis is used because payment amounts are skewed.
#
# Important: this test is exploratory. The clustering features
# are related to service volume, which may also be associated
# with payment. Interpret the result cautiously.

payment_groups = [
    group[TARGET].dropna().values
    for _, group in df.groupby("Cluster")
]

if len(payment_groups) >= 2 and all(len(g) > 0 for g in payment_groups):

    kw_statistic, kw_p_value = kruskal(*payment_groups)

    kruskal_result = {
        "Test": "Kruskal-Wallis",
        "Null_Hypothesis": (
            "Payment distributions are the same across clusters."
        ),
        "Alternative_Hypothesis": (
            "At least one cluster differs in its payment distribution."
        ),
        "Number_of_Clusters": int(len(payment_groups)),
        "Test_Statistic": float(kw_statistic),
        "P_Value": float(kw_p_value),
        "Alpha": ALPHA,
        "Reject_H0": bool(kw_p_value < ALPHA)
    }

else:

    kruskal_result = {
        "Test": "Kruskal-Wallis",
        "Status": "Not performed: insufficient cluster groups."
    }

with open(
    OUTPUT_DIR / "between_cluster_payment_test.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        kruskal_result,
        file,
        indent=2
    )

print("\nBETWEEN-CLUSTER PAYMENT TEST")
print(json.dumps(kruskal_result, indent=2))


# ================================================================
# STEP 16: VISUALIZATION 1
# SILHOUETTE SCORE BY NUMBER OF CLUSTERS
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    silhouette_df["Number_of_Clusters"],
    silhouette_df["Silhouette_Score"],
    marker="o"
)

plt.title("Silhouette Score by Number of Clusters")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.xticks(list(K_VALUES))
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "silhouette_scores.png",
    dpi=180
)

plt.show()


# ================================================================
# STEP 17: VISUALIZATION 2
# CLUSTER PROFILE COMPARISON
# ================================================================

median_features = (
    df.groupby("Cluster")[CLUSTER_FEATURES]
    .median()
)

# Log transform the medians to reduce skew.
median_features_log = np.log1p(
    median_features.clip(lower=0)
)

# Standardize each feature for visual comparison.
feature_std = median_features_log.std(ddof=0).replace(0, 1)

median_features_standardized = (
    median_features_log - median_features_log.mean()
) / feature_std

ax = median_features_standardized.T.plot(
    kind="bar",
    figsize=(11, 6)
)

ax.set_title(
    "Cluster Profiles: Standardized Log-Median Provider Characteristics"
)

ax.set_xlabel("Provider Characteristic")
ax.set_ylabel("Standardized Log-Median")
ax.legend(title="Cluster")

plt.xticks(rotation=25, ha="right")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "cluster_profile_medians.png",
    dpi=180
)

plt.show()


# ================================================================
# STEP 18: VISUALIZATION 3
# MEDICARE PAYMENT DISTRIBUTION BY CLUSTER
# ================================================================

payment_data = [
    np.log1p(
        df.loc[
            df["Cluster"] == cluster_id,
            TARGET
        ].clip(lower=0).dropna()
    )
    for cluster_id in sorted(df["Cluster"].unique())
]

cluster_labels = [
    str(cluster_id)
    for cluster_id in sorted(df["Cluster"].unique())
]

plt.figure(figsize=(10, 6))

plt.boxplot(
    payment_data,
    labels=cluster_labels,
    showfliers=False
)

plt.title(
    "Medicare Payment Distribution by Provider Cluster"
)

plt.xlabel("Cluster")
plt.ylabel("log(1 + Total Medicare Payment)")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "cluster_payment_distributions.png",
    dpi=180
)

plt.show()


# ================================================================
# STEP 19: VISUALIZATION 4
# CLUSTER SIZE DISTRIBUTION
# ================================================================

plt.figure(figsize=(8, 5))

plt.bar(
    cluster_sizes["Cluster"].astype(str),
    cluster_sizes["Provider_Count"]
)

plt.title("Number of Providers in Each Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Providers")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "cluster_sizes.png",
    dpi=180
)

plt.show()


# ================================================================
# STEP 20: SAVE REPRODUCIBILITY MANIFEST
# ================================================================

valid_test_count = int(valid_tests.sum())

significant_test_count = 0

if valid_tests.any():

    significant_test_count = int(
        (
            hypothesis_df.loc[
                valid_tests,
                "Holm_Adjusted_P_Value"
            ] < ALPHA
        ).sum()
    )

manifest = {
    "project": (
        "Predicting Medicare Payment Costs and Identifying "
        "High-Cost Provider Patterns Using Machine Learning"
    ),
    "input_file": str(INPUT_FILE),
    "original_rows_loaded": int(original_row_count),
    "rows_with_target_before_sampling": int(rows_before_sampling),
    "analysis_sample_rows": int(len(df)),
    "sample_fraction_requested": SAMPLE_FRAC,
    "maximum_sample_rows": MAX_SAMPLE_ROWS,
    "random_state": RANDOM_STATE,
    "features_used_for_clustering": CLUSTER_FEATURES,
    "target": TARGET,
    "target_used_for_clustering": False,
    "candidate_cluster_counts": list(K_VALUES),
    "silhouette_scores": {
        str(row["Number_of_Clusters"]): (
            None if pd.isna(row["Silhouette_Score"])
            else float(row["Silhouette_Score"])
        )
        for _, row in silhouette_df.iterrows()
    },
    "selected_k": int(BEST_K),
    "selected_silhouette_score": float(best_row["Silhouette_Score"]),
    "hypothesis_test": "Spearman rank correlation",
    "multiple_testing_correction": "Holm",
    "significance_level": ALPHA,
    "valid_hypothesis_tests": valid_test_count,
    "significant_tests_after_correction": significant_test_count,
    "limitations": [
        "Sample fraction is for computational feasibility, not power-based sample-size determination.",
        "Associations do not establish causation.",
        "Cluster-wise tests are exploratory.",
        "K-Means results depend on feature selection, scaling, and the chosen sample.",
        "Payment values are used for profiling and testing, not for cluster formation."
    ]
}

with open(
    OUTPUT_DIR / "clustering_diagnostics.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        manifest,
        file,
        indent=2
    )


# ================================================================
# STEP 21: DISPLAY OUTPUT FILES
# ================================================================

print("\n" + "=" * 70)
print("CLUSTERING AND HYPOTHESIS TESTING COMPLETED")
print("=" * 70)

print("\nSelected number of clusters:", BEST_K)

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nGenerated files:")

for file_path in sorted(OUTPUT_DIR.iterdir()):
    print("-", file_path.name)

print("\nInterpretation reminders:")
print("1. Cluster labels 0, 1, 2, etc. are arbitrary identifiers.")
print("2. Use the profile table to decide how to describe each cluster.")
print("3. Use Holm-adjusted p-values for significance decisions.")
print("4. Do not interpret high Medicare payment as evidence of fraud.")
print("5. Report the sample limitation and the need for power analysis.")
```
